```python
import numpy as np#lib ni import chesthunnam
```

```python
#Create marks for 3 subjects
students = np.array([  #students names ni NumPy array ga create chestundi.,   # students---mana array ki variable name.
    
    "Abhi",
    "varshini",
    "satya",
    "madhan",
    "pandu"
])

print(students)  #names display chestundi
```

```python
#Create marks for 3 subjects
marks = np.array([
    [85, 78, 92],
    [72, 88, 80],
    [90, 95, 89],
    [65, 74, 70],
    [88, 82, 91]
])

print(marks)
```

```python
#Check the shape of our marks data
print("shape:", marks.shape) #array lo rows and columns enni unnayo cheptundi.
```

```python
#Calculate Total Marks
total_marks = np.sum(marks, axis=1)   

print(total_marks)


#np.sum(marks, axis=1) → each student's 3 subject marks ni add chestundi.
#axis=1 → row-wise addition.
#total_marks → each student's total marks.
```

```python
#average_marks = np.mean(marks, axis=1)

average_marks = np.mean(marks, axis=1)

print(average_marks)

#np.mean() → average calculate chestundi.
#axis=1 → each student's 3 subjects average.
#average_marks → each student's average marks.
```

```python
#Find the Highest Total Marks

highest_total = np.max(total_marks)

print("Highest Total Marks:", highest_total)

#np.max(total_marks) → total_marks lo highest value ni find chestundi.
#highest_total → aa highest total marks ni store chestundi.
```

```python
#Find the Student with Highest Total Mark
highest_index = np.argmax(total_marks)

print("Topper:", students[highest_index])
print("Total Marks:", total_marks[highest_index])

#np.argmax(total_marks) → highest total marks unna position/index ni find chestundi.
#students[highest_index] → aa position lo unna student's name ni tiskuntundi.
#total_marks[highest_index] → aa student's total marks ni tiskuntundi.
```

```python
#Find the Lowest Total Marks
lowest_total = np.min(total_marks)

print("Lowest Total Marks:", lowest_total)

#np.min(total_marks) → total_marks lo lowest value ni find chestundi.
#lowest_total → aa value ni store chestundi
```

```python
#Find the Student with Lowest Total Marks

lowest_index = np.argmin(total_marks)

print("Lowest Scorer:", students[lowest_index])
print("Total Marks:", total_marks[lowest_index])

#np.argmin(total_marks) → lowest total marks unna index ni find chestundi.
#students[lowest_index] → aa student's name.
#total_marks[lowest_index] → aa student's total marks
```

```python
#Find the Average Marks of All Students
class_average = np.mean(marks)

print("Class Average:", class_average)

#np.mean(marks) → all students and all subjects marks ni consider chesi average calculate chestundi.
#class_average → class overall average ni store chestundi.
```

```python
#Find Highest Mark in Each Subject
highest_subject_marks = np.max(marks, axis=0)

print("Highest marks in each subject:", highest_subject_marks)

#np.max() → highest value find chestundi.
#axis=0 → column-wise calculation.
#So, each subject lo highest mark separately vastundi
```

```python
#Find Lowest Mark in Each Subject
lowest_subject_marks = np.min(marks, axis=0)

print("Lowest marks in each subject:", lowest_subject_marks)

#np.min() → lowest value find chestundi.
#axis=0 → column-wise calculation.
#So, each subject lo lowest mark separately vastundi
```
